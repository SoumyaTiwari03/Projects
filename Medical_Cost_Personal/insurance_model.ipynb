```python
import pandas as pd
import pickle

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import r2_score

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
```

```python
df = pd.read_csv("insurance.csv")

print(df.head())
```

```python
print(df.shape)

df.info()
```

```python
print(df.isnull().sum())
```

```python
df.describe()
```

```python
print(df["sex"].unique())

print(df["smoker"].unique())

print(df["region"].unique())
```

```python
X = df.drop("charges", axis=1)

y = df["charges"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=0
)
```

```python
X_train.sample(3)
```

```python
X_test.sample(3)
```

```python
y_train.sample(3)
```

```python
y_test.sample(3)
```

```python
categorical_features = [
    "sex",
    "smoker",
    "region"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)
```

```python
#Linear Regression

lr_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

lr_pipeline.fit(X_train, y_train)

lr_pred = lr_pipeline.predict(X_test)

print(
    "Linear Regression R2 Score:",
    r2_score(y_test, lr_pred)
)
```

Accuracy From Linear Regression : 79.99%


```python
#Decision Tree

dt_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeRegressor(
        random_state=0
    ))
])

dt_pipeline.fit(X_train, y_train)

dt_pred = dt_pipeline.predict(X_test)

print(
    "Decision Tree R2 Score:",
    r2_score(y_test, dt_pred)
)
```

Accuracy From Decision Tree : 71.42%

```python
#Random Forest

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        random_state=0
    ))
])

rf_pipeline.fit(X_train, y_train)

rf_pred = rf_pipeline.predict(X_test)

print(
    "Random Forest R2 Score:",
    r2_score(y_test, rf_pred)
)
```

Accuracy From Random Forest : 87.74%

```python
#SVM

svm_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", SVR())
])

svm_pipeline.fit(X_train, y_train)

svm_pred = svm_pipeline.predict(X_test)

print(
    "SVM R2 Score:",
    r2_score(y_test, svm_pred)
)
```

Accuracy From SVM : 10.39%

```python
#Random Forest Pipeline is taken since it gives the most accuracy of all.
pickle.dump(
    rf_pipeline,
    open(
        "insurance_model.pkl",
        "wb"
    )
)

print("Model Saved")
```
